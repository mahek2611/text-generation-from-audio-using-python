```python
import whisper
from pydub import AudioSegment
from pydub.utils import make_chunks
import re

# Load the Whisper model
model = whisper.load_model("medium")

# Load the audio file
audio = AudioSegment.from_file("/content/drive/MyDrive/audio_file.mp3")

# Define chunk size (in milliseconds, e.g., 30 seconds = 30000 ms)
chunk_size = 30000
chunks = make_chunks(audio, chunk_size)

# Store transcriptions
transcriptions = []

# Transcribe each chunk
for i, chunk in enumerate(chunks):
    chunk_name = f"chunk{i}.mp3"
    chunk.export(chunk_name, format="mp3")  # Export the chunk to a temporary file
    print(f"Processing chunk {i + 1}/{len(chunks)}...")

    # Transcribe the chunk
    result = model.transcribe(chunk_name)
    transcriptions.append(result["text"])

# Combine all transcriptions into one text
final_transcription = " ".join(transcriptions)

# Split the text into sentences
sentences = re.split(r'(?<=[.!?])\s+', final_transcription)  # Split at sentence boundaries

# Format as plain text
print("\nAccurate Transcription:\n")
for sentence in sentences:
    if sentence.strip():  # Skip empty lines
        print(sentence.strip())
```

```python
import whisper
from pydub import AudioSegment
from pydub.utils import make_chunks
import re

# Load the Whisper model
model = whisper.load_model("large")

# Load the audio file
audio = AudioSegment.from_file("/content/drive/MyDrive/audio_file.mp3")

# Define chunk size (in milliseconds, e.g., 30 seconds = 30000 ms)
chunk_size = 30000
chunks = make_chunks(audio, chunk_size)

# Store transcriptions
transcriptions = []

# Transcribe and translate each chunk
for i, chunk in enumerate(chunks):
    chunk_name = f"chunk{i}.mp3"
    chunk.export(chunk_name, format="mp3")  # Export the chunk to a temporary file
    print(f"Processing chunk {i + 1}/{len(chunks)}...")

    # Transcribe and translate the chunk to English
    result = model.transcribe(chunk_name, task="translate")
    transcriptions.append(result["text"])

# Combine all translations into one text
final_transcription = " ".join(transcriptions)

# Split the text into sentences
sentences = re.split(r'(?<=[.!?])\s+', final_transcription)  # Split at sentence boundaries

# Print the final transcript in English
print("\nFormatted English Transcript:\n")
for sentence in sentences:
    if sentence.strip():  # Skip empty lines
        print(sentence.strip())
```
